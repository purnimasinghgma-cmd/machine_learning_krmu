```python
import pandas as pd
```

```python
df = pd.read_csv("insurance.csv")
df
```

```python
df.info()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(sparse_output = False) #here we are  creating the object
data = encoder.fit_transform(df[["sex"]]) #[[]] means dataframe
cols = encoder.get_feature_names_out(["sex"])
new_data = pd.DataFrame(data,columns = cols)
new_data
```

```python

df.drop(["sex", "smoker"],axis =1, inplace = True)
df
```

```python
final_data = pd.concat([df, new_data], axis =1)
final_data
```
