```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate Q1 and Q3
Q1 = df["SepalLengthCm"].quantile(0.25)
Q3 = df["SepalLengthCm"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Find outliers
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

out = df[
    (df["SepalLengthCm"] < lower) |
    (df["SepalLengthCm"] > upper)
]["SepalLengthCm"]

plt.figure(figsize=(5, 8))

sns.boxplot(
    y=df["SepalLengthCm"],
    color="lightblue"
)
plt.scatter(
    [0] * len(out),
    out,
    color="red",
    s=50,
    zorder=5,
    label="Outliers"
)

plt.title("Box Plot of Sepal Length with Custom Outliers")
plt.ylabel("Sepal Length (cm)")
plt.legend()
plt.show()
```
