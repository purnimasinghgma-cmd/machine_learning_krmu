```python
import numpy as np
```

```python
X = np.array([[40,5],
              [30,6],
              [35,5.6],
              [120,25],
              [110,35],
              [140,45]])
label=np.array([1,1,1,0,0,0])
label
```

```python
data = [45,10]
X[0]
```

```python
data = [45,10]
distance = np.array([np.sqrt(np.sum(x-data)**2) for x in X])
distance
```

```python
d = np.argsort(distance) #argshort gives the index of sortesd array in order
d
```

```python
d = np.argsort(distance)[:3] # gives the index of the 3 nearest neighbors
label[d]
```
