```python
import pandas as pd
```

```python
data = {
    "Name": ["radhe","shyam"],
    "City": ["odisha","kendrapara"],
    "Unit": [45,10],
    "Sp": [3400,5600],
    "Cp": [3000,4200]
}
df = pd.DataFrame(data)
df
```

```python
df["Total_cost"]=df["Sp"]*df["Cp"]
df
```

```python
df["Diff"]=df["Sp"]-df["Cp"]
df
```

```python
df["Prof_Loss"]=df["Diff"].apply(lambda x: "Profit" if x>0 else "Loss")
df
```

```python
df["Order_Type"]=df["Unit"].apply(lambda x: "Bulk_Order" if x>20 else "Normal_Order")
```

```python
df["Profit_Loss"] = ["profit" if x>0 else "Loss" for x in df["Diff"]]
df
```
